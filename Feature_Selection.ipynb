### **CleanUP**

```python
x=df.drop(['Pass/Fail',"Time"],axis=1)
y=df['Pass/Fail']
```

```python
x.shape
```

### Removing Constant Value/Columns

```python
constant_cols=x.columns[x.nunique() <= 1]
len(constant_cols)
```

```python
x=x.drop(columns=constant_cols)
x.shape
```

### Removing DuplicateColumns

```python
duplicate_columns = x.T.duplicated()
print("Number of duplicated columns:", duplicate_columns.sum())
```

## Missing values
we remove the columns if they have more than 50% of null values

```python
missing = x.isnull().sum()

missing[missing > 0].sort_values(ascending=False).head(20)
```

```python
missing_percentage=x.isnull().mean()*100
missing_percentage.sort_values(ascending=False).head()
```

```python
features_to_drop = missing_percentage[missing_percentage > 50].index   #28 columns
x=x.drop(columns=features_to_drop)
x.shape
```

Data Split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

x_train = imputer.fit_transform(x_train)
x_test = imputer.transform(x_test)
```

```python
import numpy as np

print(np.isnan(x_train).sum())
print(np.isnan(x_test).sum())
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model.fit(x_train, y_train)
```

```python
y_pred = log_model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

###**NOW FeatureSelection on Training dataset to avoid leakage**

ANOVA / F-test

```python
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
selector = SelectKBest(score_func=f_classif, k=40)

x_train_selected = selector.fit_transform(x_train, y_train)
x_test_selected = selector.transform(x_test)
```

```python
print(x_train_selected.shape)
print(x_test_selected.shape)
```

```python
log_model_fs = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model_fs.fit(x_train_selected, y_train)
```

```python
y_pred_fs = log_model_fs.predict(x_test_selected)
```

```python
from sklearn.metrics import confusion_matrix

cm_fs = confusion_matrix(y_test, y_pred_fs)
print(cm_fs)
```

```python
from sklearn.metrics import balanced_accuracy_score

balanced_accuracy = balanced_accuracy_score(y_test, y_pred_fs)

ber = 1 - balanced_accuracy

print("Balanced Accuracy:", balanced_accuracy)
print("BER:", ber)
```

**T-Test**

```python
from scipy.stats import ttest_ind
import numpy as np
x_pass = x_train[y_train == -1]
x_fail = x_train[y_train == 1]
t_scores = []

for i in range(x_train.shape[1]):
    t_stat, p_value = ttest_ind(
        x_pass[:, i],
        x_fail[:, i],
        equal_var=False
    )
    t_scores.append(abs(t_stat))

t_scores = np.array(t_scores)

top_40_indices = np.argsort(t_scores)[-40:]
x_train_ttest = x_train[:, top_40_indices]
x_test_ttest = x_test[:, top_40_indices]
print(x_train_ttest.shape)
print(x_test_ttest.shape)
```

```python
log_model_ttest = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_model_ttest.fit(x_train_ttest, y_train)
y_pred_ttest = log_model_ttest.predict(x_test_ttest)
cm_ttest = confusion_matrix(y_test, y_pred_ttest)
print(cm_ttest)
```

```python
balanced_accuracy_ttest = balanced_accuracy_score(y_test, y_pred_ttest)

ber_ttest = 1 - balanced_accuracy_ttest

print("Balanced Accuracy:", balanced_accuracy_ttest)
print("BER:", ber_ttest)
```

Quasi Constant

```python
from scipy.stats import mode

most_common_percentage = []

for i in range(x_train.shape[1]):
    most_common_count = mode(x_train[:, i], keepdims=True).count[0]
    percentage = most_common_count / len(x_train)
    most_common_percentage.append(percentage)
most_common_percentage = np.array(most_common_percentage)

quasi_constant_features = np.where(
    most_common_percentage >= 0.99
)[0]

print("Quasi-constant features:", len(quasi_constant_features))
```

```python
x_train_qc = np.delete(x_train, quasi_constant_features, axis=1)
x_test_qc = np.delete(x_test, quasi_constant_features, axis=1)
```

```python
print(x_train_qc.shape)
print(x_test_qc.shape)
```

**CoRelation**

```python
import pandas as pd

x_train_df = pd.DataFrame(x_train_qc)
corr_matrix = x_train_df.corr().abs()
```

```python
high_corr = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr_pairs = high_corr.stack()

high_corr_pairs = high_corr_pairs[
    high_corr_pairs > 0.90
]

print(high_corr_pairs)
```

```python
print("Number of highly correlated pairs:", len(high_corr_pairs))
```

```python
correlated_pairs = high_corr_pairs.reset_index()
correlated_pairs.columns = ["Feature_1", "Feature_2", "Correlation"]

print(correlated_pairs.head(20))
```

```python
unique_correlated_features = set(
    correlated_pairs["Feature_1"]
).union(
    set(correlated_pairs["Feature_2"])
)

print("Unique correlated features:", len(unique_correlated_features))
```

**S2N**

```python
import numpy as np

x_pass = x_train_qc[y_train == -1]
x_fail = x_train_qc[y_train == 1]

mean_pass = np.mean(x_pass, axis=0)
mean_fail = np.mean(x_fail, axis=0)

std_pass = np.std(x_pass, axis=0)
std_fail = np.std(x_fail, axis=0)

s2n_scores = (
    np.abs(mean_pass - mean_fail)
    / (std_pass + std_fail)
)
top_40_s2n = np.argsort(s2n_scores)[-40:]
x_train_s2n = x_train_qc[:, top_40_s2n]
x_test_s2n = x_test_qc[:, top_40_s2n]
print(x_train_s2n.shape)
print(x_test_s2n.shape)
```

```python
log_model_s2n = LogisticRegression(max_iter=1000, random_state=42)

log_model_s2n.fit(x_train_s2n, y_train)

y_pred_s2n = log_model_s2n.predict(x_test_s2n)
cm_s2n = confusion_matrix(y_test, y_pred_s2n)

print(cm_s2n)
```

```python
balanced_accuracy_s2n = balanced_accuracy_score(y_test, y_pred_s2n)
ber_s2n = 1 - balanced_accuracy_s2n

print("S2N Balanced Accuracy:", balanced_accuracy_s2n)
print("S2N BER:", ber_s2n)
```
